In churn prediction ,recall (and metrics derived from it,like F1) is usually  the most important -here why!
1. Business context Churn prediction is assymetric in cost:
- False Negative(FN):You predict a customer will not churn,but they actually leave ->you miss chance to intervene and keep them.
- False positive (FP)-You predict a customer will churn,but they stay and youu spend on some unnecessary retention resources.
- Missing Churners(FN) - usually cost more than wrongly targeting a loyal customer.

Typical Priority
- If retention campgains are cheap -> prioritize recall( catch every possible churner)
- If retention are expensive ->balance precision  and recall using F1 score or a precision trade off
- If business want a ranking of churn risk - Use ROC-AUC or PR-AUC score to evaluate the model

# Model Building

```python
# Divide the col into X and y
X=df.drop(columns=['Churn'])
y=df['Churn']

# Scale the Data 
num_cols=['tenure','MonthlyCharges','TotalCharges']
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X[num_cols]=scaler.fit_transform(X[num_cols])


# Train Test Split
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import  lightgbm as  lgb
from sklearn.metrics import accuracy_score,classification_report
import time
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=24,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score,recall_score,precision_score,f1_score
models={
    "Random Forest":RandomForestClassifier(n_estimators=300,min_samples_leaf=5,class_weight='balanced'),
    "Logistic Regression":LogisticRegression(max_iter=2000,class_weight='balanced'),
    "LIGHT GBM":lgb.LGBMClassifier(
        n_estimators=300,learning_rate=0.05,max_depth=4,class_weight='balanced',verbose=-1),
}

result=[]
for name,model in models.items():
    start=time.time()
    model.fit(X_train,y_train)
    proba=model.predict_proba(X_test)[:,1]
    pred=(proba>=0.5).astype(int)

    result.append({
        "model":name,
        "accuracy_score":accuracy_score(y_test,pred),
        "roc_auc":roc_auc_score(y_test,proba),
        "recall":recall_score(y_test,pred),
        "precision":precision_score(y_test,pred),
        "f1":f1_score(y_test,pred),
        "seconds":round(time.time()-start,2),

    })
    print(f"\n{name}")
    print(classification_report(y_test,pred,target_names=["Stay","Churn"]))
pd.DataFrame(result).round(3)



```

```python
from xgboost import XGBClassifier

# XGBoost has no class_weight, so use scale_pos_weight (stay count / churn count)
spw = (y_train == 0).sum() / (y_train == 1).sum()

xgb = XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=4,
    scale_pos_weight=spw, eval_metric='logloss',
    random_state=24, n_jobs=-1
)

start = time.time()
xgb.fit(X_train, y_train)
proba = xgb.predict_proba(X_test)[:, 1]
pred = (proba >= 0.5).astype(int)

result.append({
    "model": "XGBoost",
    "accuracy_score": accuracy_score(y_test, pred),
    "roc_auc": roc_auc_score(y_test, proba),
    "recall": recall_score(y_test, pred),
    "precision": precision_score(y_test, pred),
    "f1": f1_score(y_test, pred),
    "seconds": round(time.time() - start, 2),
})

pd.DataFrame(result).round(3)
```

# Hyper Paramter Tuning (Optuna) and Experimental Tracking (ML Flow)

import mlfl

```python
import mlflow,optuna
from sklearn.model_selection import StratifiedKFold,cross_val_score

mlflow.set_experiment("Telecom_Chur")
cv=StratifiedKFold(5,shuffle=True,random_state=24)

def objective(trial):
    params={
        "n_estimators":trial.suggest_int("n_estimators",100,500),
        "learning_rate":trial.suggest_float("learning_rate",0.01,0.2,log=True),
        "max_depth":trial.suggest_int("max_depth",2,6),
        "min_child_samples":trial.suggest_int("min_child_samples",10,80),
        "reg_lambda":trial.suggest_float("reg_lambda",0.001,10,log=True)
    }
    model=lgb.LGBMClassifier(**params,class_weight="balanced",random_state=24,verbose=-1)
    score=cross_val_score(model,X_train,y_train,cv=cv,scoring='roc_auc',n_jobs=-1).mean()
    with mlflow.start_run(nested=True): 
        mlflow.log_params(params)
        mlflow.log_metric("cv_roc_score",score)
    return score
with mlflow.start_run(run_name="lightgbm-optuna"):
    study=optuna.create_study(direction="maximize")
    study.optimize(objective,n_trials=15)
    mlflow.log_metric(  "best_score",study.best_value)

print("Best Score:",round(study.best_value,3))
print("Best Params:",study.best_params)
```

```python
best = lgb.LGBMClassifier(**study.best_params, class_weight="balanced",
                          random_state=24, verbose=-1)
best.fit(X_train, y_train)
proba = best.predict_proba(X_test)[:, 1]
print("Tuned LightGBM test ROC-AUC:", round(roc_auc_score(y_test, proba), 3))
```
